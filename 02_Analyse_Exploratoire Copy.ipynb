```python
from pyspark.sql import functions as F
from pyspark.sql.types import *

# 1. Création de données fictives
data = [
    ("2024-01-01", "Basic", 10, 150),
    ("2024-01-02", "Premium", 5, 250),
    ("2024-02-01", "Basic", 12, 180),
    ("2024-02-15", "Premium", 8, 400),
    ("2024-03-01", "Ultra", 3, 600),
    ("2024-03-20", "Basic", 15, 225)
]

columns = ["Date", "Plan", "Quantity", "Revenue"]
df = spark.createDataFrame(data, schema=columns)

# 2. Transformation : Ajouter une colonne de taxe (20%)
df_final = df.withColumn("Revenue_with_Tax", F.col("Revenue") * 1.20)

# 3. Afficher le résultat
display(df_final)
```

```python
# On enregistre le résultat dans le Lakehouse sous forme de table
df_final.write.format("delta").mode("overwrite").saveAsTable("Fact_Sales_Exploratory")
```
